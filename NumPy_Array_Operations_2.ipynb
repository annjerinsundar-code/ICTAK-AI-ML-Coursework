Create a NumPy array with numbers from 1 to 20.
Select every 3rd element starting from the 2nd element and store them in a new array.

```python
import numpy as np

# NumPy array with numbers 1 to 20
arr = np.arange(1, 21)
""" numpy.arange() function creates an array of evenly spaced values within a given interval.
It is similar to Python's built-in range() function but returns a NumPy array instead of a list. """

# Print original array
print(f"Original array:\n{arr}")

# Select every 3rd element starting from the 2nd element
updated_arr = arr[1::3]

print(f"\nUpdated array:\n{updated_arr}")
```
