Load the dataset Iris data

o Display the first 10 rows and basic statistics.

o Filter all rows where sepal_length > median and species is "versicolor"
.

o Group the data by species and compute the mean petal_length.

```python
import seaborn as sns
import pandas as pd

# Load Seaborn dataset - Iris
df = sns.load_dataset('iris')

# Print first 10 entries
print(f"Iris Dataset - First 10 Rows\n\n{df.head(10)}")

# Print basic statistics
print(f"\n\n\nIris Dataset - Basic Statistics\n\n{df.describe()}")
```

```python
# Find median of sepal_length
sepal_median = df["sepal_length"].median()
print(f"Median of sepal_length = {sepal_median}")

# Filter all rows where sepal_length > median and species is "versicolor"
df[(df['sepal_length'] > sepal_median) & (df['species'] == 'versicolor')]
```

```python
# Group the data by species and compute the mean petal_length.

petal_length_mean = df.groupby('species')['petal_length'].mean()
print("Mean of Petal length by Species:\n\n ",petal_length_mean)
```
