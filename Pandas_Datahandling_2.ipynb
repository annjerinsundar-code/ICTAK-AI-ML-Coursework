Load the dataset Iris data.

Add a new column named petal_ratio = petal_length / petal_width.

o Find the top 5 rows with the highest petal_ratio.

```python
import seaborn as sns
import pandas as pd

# Load seaborn dataset - Iris
df = sns.load_dataset('iris')

# Add a new column named petal_ratio = petal_length / petal_width
df['petal_ratio'] = [df['petal_length'][i] / df['petal_width'][i] for i in range(len(df))]

# Sort data by petal_ratio and print the first 5 entries
df_sorted = df.sort_values('petal_ratio',ascending=False)
df_sorted.head(5)
```
