A list contains the names of passengers in order of booking. A cancellation occurs; remove
the name from the list and shift the rest. Also, insert a new booking at a specified position.

**Sample Input:**
Booked = ['Amit', 'Sara', 'John'],
Cancel = 'Sara',
New = 'Neha', Position = 1

**Sample Output:**
['Amit', 'Neha', 'John']

```python
# Input the list of passengers separated by commas
booked = input("Enter the passengers names (separated by commas only): ").split(',')

# Input the passenger's name to be cancelled
cancel = input("Enter the passenger to be cancelled: ")

# Find the passenger's index position from the list
position = booked.index(cancel)

# Remove the cancelled passenger
booked.remove(cancel)

# Input the new passenger's name to be inserted
new = input("Enter the new passenger: ")

# Insert the new passenger at the cancelled passenger's index
booked.insert(position, new)

print("\n\nPassenger List: ",booked)
```
