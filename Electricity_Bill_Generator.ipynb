Create a bill calculator where the user enters the number of units consumed. The rate is
₹5 per unit for the first 100 units, ₹7 for the next 100, and ₹10 for anything above 200
units. Calculate and display the total bill.
Sample Input:
Units consumed = 275
Sample Output:
Bill amount = ₹1000 (100×5 + 100×7 + 75×10)

```python
# Get the number of units
units = int(input("Enter the number of units consumed: "))

# Initialize bill_amount
bill_amount = 0

# Check for a non zero unit value
if units <0:
  print("Invalid unit entered")

# Check the required rate for calculating bill amount
else:
  if units <= 100:
    bill_amount = units * 5
  elif units <= 200:
    bill_amount = (100 * 5) + (units - 100) * 7
  else:
    bill_amount = (100 * 5) + (100 * 7) + (units - 200) * 10
  print(f"Bill amount: ₹{bill_amount}")
```
