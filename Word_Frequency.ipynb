Write a program that takes a sentence and displays the frequency of each word using a
dictionary.

**Sample Input:**

Sentence: hello world hello python

**Sample Output:**

hello: 2
world: 1
python: 1

```python
# Input the sentence from the user; Convert each word to lowercase
sentence = input("Enter the sentence: ").lower()

# Create an empty dictionary
word_frequency = {}

# Iterate each word in the sentence
for word in sentence.split():
  # Assigns the key-value pairs; If a word is present in the sentence, the get()
  # returns 1, otherwise the default 0
  word_frequency[word] = word_frequency.get(word,0)+1

# Print the dictionary in the required format
for word,frequency in word_frequency.items():
  print(f"{word}:{frequency}")
```
