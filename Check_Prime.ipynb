Write a function that checks whether a number is prime. Call the function with user input.

**Sample Input:**
 Enter a number : 17

**Sample output:**
17 is a prime number.

```python
def check_prime(number):
  """ Function to check whether a given integer is prime or not."""

  # Numbers: 0 and 1 are not prime
  if(number < 2):
    return False

  # Check whether the number has any factors within the given range
  for i in range(2,number):
    if number % i == 0:
      return False

  # Returns true if both the conditions are not met
  return True
```

```python
# User input number
num = int(input("Enter a number: "))

# Check whether the given number is prime using the check_prime() function
if check_prime(num):
  print(f"{num} is a prime number")
else:
  print(f"{num} is not a prime number")
```
