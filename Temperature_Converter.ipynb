Write a Python program to convert a given temperature value between Celsius and
Fahrenheit. Accept input as a number and a unit indicator ('C' or 'F').
Sample Input:
Temp = "37"
Unit = "C"
Sample Output:
Temperature in Fahrenheit: 98.6°F

Celsius → Fahrenheit: F = (C * 9/5) + 32

Fahrenheit → Celsius:
C = ( F − 32 ) * 5/9

```python
# Enter the temperature value
temp = float(input("Enter the temperature: "))
```

```python
# Enter the temperature unit value
unit = input("Enter the unit (C or F): ")
```

```python
# Convert unit values to uppercase; check and calculate the required temperature value
if unit.upper() ==  "C":
  fahrenheit = (temp * 9/5) + 32
  print("Temperature in Fahrenheit:", fahrenheit, "°F")
elif unit.upper() == "F":
  celsius = (temp - 32) * 5/9
  print("Temperature in Celsius:", celsius, "°C")
else:
  print("Invalid unit")
```
