Using Matplotlib, create:

o A histogram of sepal_length with 15 bins.

o A scatter plot of sepal_length vs petal_length, color-coded by species.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the iris dataset
df = sns.load_dataset("iris")

# Plot customized Sepal Length histogram using matplotlib
plt.hist(df['sepal_length'], bins=15, edgecolor="b",color="orange")

# x axis labelling
plt.xlabel('Sepal Length')

# y axis labelling
plt.ylabel('Frequency')

# histogram title
plt.title('Histogram of Sepal Length')

# Print histogram
plt.show()

```

```python
# Separate the species
species = df['species'].unique()

# Plot the scatter plot
for s in species:
    df_species = df[df['species'] == s]
    plt.scatter(df_species['sepal_length'], df_species['petal_length'], label=s)

plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.title('Scatter Plot of Sepal Length vs Petal Length, Color coded by Species')
plt.legend()
plt.show()

```
