Write a program that takes input of 3 students’ names and their marks, stores them in a
dictionary, and prints the average marks.

**Sample Input:**

Enter name: Alice;
Enter marks: 80;
Enter name: Bob;
Enter marks: 90;
Enter name: Charlie;
Enter marks: 70

**Sample Output:**

Average Marks: 80.0

```python
# Create an empty dictionary for student names and marks
students = {}

# Input the names and marks of 3 students
for i in range(3):
    name = input("Enter name: ").capitalize #capitalizes the names
    marks = float(input("Enter marks: "))   #marks typecasted to float
    students[name] = marks #add name and marks to dictionary

# Compute average marks and print
average_marks = sum(students.values()) / len(students)
print("\n\nAverage Marks:", average_marks)
```
