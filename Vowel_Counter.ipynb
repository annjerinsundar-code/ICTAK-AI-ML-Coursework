Write a function that takes a string and returns the number of vowels in it.

**Sample Input:**
Enter a string: Python Programming

**Sample output:**
Number of vowels: 4

```python
def vowel_count(text):
  """ Function to check the number of vowels in an input string """
  count = 0
  vowels = "aeiou"

  # Iterate through each character in lowercase
  for char in text.lower():
    # Increment the count if the character is a vowel
    if char in vowels:
      count += 1
  return count
```

```python
# User input string
text = input("Enter a string: ")

# Call function inside print() to print the vowel count
print("Number of vowels: ", vowel_count(text))
```
