Write a program that takes two numbers as input: a range limit N and a divisor D. Count
how many numbers between 1 and N are divisible by D.

**Sample Input:**

Enter the range limit (N): 50
Enter the divisor (D): 5

**Sample output:**

Count of numbers divisible by 5: 10
(i.e., 5, 10, 15, ..., 50)

```python
# Get the integers: range limit and divisor
N = int(input("Enter the range limit: "))
D = int(input("Enter the divisor: "))

# Initialize count to 0
count = 0

# List for storing divisors
divisible_numbers = []

# Count numbers divisible by D within the range 1 - N
for i in range(1, N + 1):
    if i % D == 0:
        count += 1
        divisible_numbers.append(i)

# Display the result
print(f"\nCount of numbers divisible by {D}: {count}")
print("Divisible numbers: ",divisible_numbers)
```
