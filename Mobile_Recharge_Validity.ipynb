Write a program that takes the recharge date and today’s date as input and checks
whether the 28-day recharge plan is still valid or expired.

Sample Input:
Recharge date: 2025-07-01
Today : 2025-07-30

Sample Output:
Plan expired. Recharge again.

```python
from datetime import datetime

# Get the recharge date and today's date
recharge_date = input("Enter the recharge date (YYYY-MM-DD): ")
today_date = input("Enter today's date (YYYY-MM-DD): ")
```

```python
# Convert strings to dates
recharge_date = datetime.strptime(recharge_date, "%Y-%m-%d")
today_date = datetime.strptime(today_date, "%Y-%m-%d")
```

```python
# Calculate the difference in days
days_difference = (today_date - recharge_date).days
```

```python
# Check 28 day plan validity
if days_difference < 28:
  print(f"Plan is valid for {28 - days_difference} more days.")
elif  days_difference == 28:
  print("Plan will expire tomorrow.")
else:
  print("Plan expired. Recharge again.")
```
