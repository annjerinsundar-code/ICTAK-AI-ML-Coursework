Using Seaborn, plot:

o A pairplot of the entire dataset.

o A boxplot showing petal_width distribution for each species.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load Iris dataset to 'iris'
iris = sns.load_dataset("iris")

# Plot pairplot of iris with different colours for each species
sns.pairplot(iris, hue="species")
plt.title("Pairplot of Iris Dataset")
plt.show()
```

```python
# Plot boxplot of petal_width for each species
sns.boxplot(x="species", y="petal_width", hue = 'species', data=iris)
plt.title("Petal Width Distribution by Species")
plt.show()
```
