Write a program that takes the user's account balance and withdrawal amount. If
sufficient balance is available, deduct the amount and display the updated balance.
Otherwise, print an error message.



Sample Input:
Balance = 5000
Withdrawal = 1500



Sample Output:
Transaction successful.
Remaining balance: ₹3500

```python
#Input the account balance from the user. Int values will be automatically typecasted to float.
balance = float(input("Enter the balance amount: "))

#Input the withdrawal amount from the user. Int values will be automatically typecasted to float.
withdrawal_amount = float(input("Enter the withdrawal amount: "))

#Check whether sufficient balance is available
#If withdrawal_amount is greater than balance: print Invalid Message
if withdrawal_amount > balance:
    print("Insufficient balance.")

#If withdrawal_amount is lesser than balance: print Success Message and balance amount
else:
    balance -= withdrawal_amount
    print("\nTransaction successful.")
    print(f"Remaining balance: ₹{balance}")
```
