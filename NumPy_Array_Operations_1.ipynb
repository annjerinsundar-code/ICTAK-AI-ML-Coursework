Create a NumPy array of shape (5, 5) with random integers between 1 and 50.

o Find the row with the maximum sum.

o Replace all elements greater than the mean with 0.

```python
import numpy as np

# Create a 5 × 5 array with random integers from 1 to 50
arr = np.random.randint(1, 51, size=(5, 5))

print("Original Array:")
print(arr)

# Find the sum of each row
row_sums = np.sum(arr, axis=1)

# Find the row with the maximum sum
max_row = np.argmax(row_sums)

print("\nRow sums:")
print(row_sums)

print("\nRow with maximum sum:")
print(arr[max_row])

# Calculate the mean of the array
mean = np.mean(arr[max_row])

print("\nMean of row with maximum sum:", mean)

# Replace all elements greater than the mean, with 0, in the original array
arr[arr > mean] = 0

print("\nArray after replacing elements greater than mean with 0:")
print(arr)
```
